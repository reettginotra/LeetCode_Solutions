```python
import pandas as pd
import os
from glob import glob

# ==========================
# Specify folder paths
# ==========================
folder1 = "/content/recommissioned_batteries"
folder2 = "/content/regular_alt_batteries"
folder3 = "/content/second_life_batteries"

folders = {
    "folder1": folder1,
    "folder2": folder2,
    "folder3": folder3
}

# ==========================
# Merge CSVs in each folder
# ==========================
for folder_name, folder_path in folders.items():

    csv_files = glob(os.path.join(folder_path, "*.csv"))

    print(f"\nProcessing {folder_name}")
    print(f"Found {len(csv_files)} CSV files")

    df_list = []

    for file in csv_files:
        try:
            df = pd.read_csv(file)
            df_list.append(df)
            print(f"Loaded: {os.path.basename(file)}")
        except Exception as e:
            print(f"Error reading {file}: {e}")

    if len(df_list) > 0:
        merged_df = pd.concat(df_list, ignore_index=True)

        output_file = f"/content/merged_{folder_name}.csv"
        merged_df.to_csv(output_file, index=False)

        print(f"Saved: {output_file}")
        print(f"Shape: {merged_df.shape}")
    else:
        print("No valid CSV files found.")
```

```python
import pandas as pd

recommended = pd.read_csv("/content/merged_folder1.csv")
regular = pd.read_csv("/content/merged_folder2.csv")
second_life = pd.read_csv("/content/merged_folder3.csv")

print(recommended.shape)
print(regular.shape)
print(second_life.shape)
```

```python
recommended["battery_class"] = 0
regular["battery_class"] = 1
second_life["battery_class"] = 2

battery_map = {
    0:"Recommended",
    1:"Regular",
    2:"Second_Life"
}
```

```python
df = pd.concat(
    [recommended, regular, second_life],
    ignore_index=True
)

print(df.shape)

df.head()
```

```python
df.info()

df.describe(include='all')
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

sns.heatmap(
    df.isnull(),
    cbar=False
)

plt.title("Missing Values")
plt.show()
```

```python
df.drop_duplicates(inplace=True)

df.reset_index(drop=True,inplace=True)

df.isnull().sum()
```

```python
df['start_time'] = pd.to_datetime(
    df['start_time'],
    errors='coerce'
)

df['time'] = pd.to_numeric(
    df['time'],
    errors='coerce'
)
```

```python
df['hour'] = df['start_time'].dt.hour

df['day'] = df['start_time'].dt.day

df['month'] = df['start_time'].dt.month

df['weekday'] = df['start_time'].dt.weekday
```

```python
df['voltage_diff'] = df['voltage_charger'].diff()

df['voltage_rolling_mean'] = (
    df['voltage_charger']
    .rolling(10)
    .mean()
)

df['voltage_rolling_std'] = (
    df['voltage_charger']
    .rolling(10)
    .std()
)
```

```python
df['temp_diff'] = (
    df['temperature_battery']
    .diff()
)

df['temp_rolling_mean'] = (
    df['temperature_battery']
    .rolling(10)
    .mean()
)

df['temp_rolling_std'] = (
    df['temperature_battery']
    .rolling(10)
    .std()
)
```
